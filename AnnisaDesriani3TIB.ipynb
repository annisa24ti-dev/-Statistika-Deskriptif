```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(
    '/content/annisa_statistika_deskriptif_python.xlsx',
    sheet_name='Data'
)

display(df)
print("Jumlah mahasiswa:", len(df))
```

```python
#tabel frekuensi
urutan_grade = ["A", "AB", "B", "BC", "C", "F"]

frekuensi = df["Grade"].value_counts().reindex(urutan_grade).fillna(0).astype(int)
persentase = (frekuensi / frekuensi.sum() * 100).round(2)

tabel_frekuensi = pd.DataFrame({
    "Frekuensi": frekuensi,
    "Persentase (%)": persentase
})

display(tabel_frekuensi)
```

```python
# grafik garis
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(df) + 1),
    df["Final Score"],
    marker="o",
    linewidth=2,
    markersize=5
)

plt.title("Grafik Garis Nilai Akhir Mahasiswa", fontsize=13, fontweight="bold")
plt.xlabel("Urutan Mahasiswa")
plt.ylabel("Nilai Akhir")

plt.xticks(range(1, len(df) + 1))
plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
#grafik batang
plt.figure(figsize=(8, 4))

warna = ['#4285F4', '#34A853', '#FBBC05', '#EA4335', '#9C27B0', '#FF7043']

bars = plt.bar(
    urutan_grade,
    frekuensi.values,
    color=warna
)

plt.title("Distribusi Grade Mahasiswa", fontsize=13, fontweight="bold")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")

for bar, nilai_bar in zip(bars, frekuensi.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        nilai_bar + 0.2,
        str(nilai_bar),
        ha="center",
        fontweight="bold"
    )

plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
#pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    frekuensi.values,
    labels=urutan_grade,
    autopct="%1.1f%%",
    startangle=90,
    colors=warna,
    explode=[0.03] * len(urutan_grade)
)

plt.title("Persentase Grade Mahasiswa", fontsize=13, fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
#statistika deskriptif
nilai = df["Final Score"]

statistik = pd.DataFrame({
    "Statistik": [
        "Jumlah Data",
        "Mean",
        "Median",
        "Mode",
        "Minimum",
        "Q1",
        "Q2",
        "Q3",
        "Maximum",
        "Variance",
        "Standard Deviation",
        "Standard Error",
        "Skewness",
        "Kurtosis"
    ],
    "Nilai": [
        nilai.count(),
        nilai.mean(),
        nilai.median(),
        ", ".join([f"{x:.2f}" for x in nilai.mode()]),
        nilai.min(),
        nilai.quantile(0.25),
        nilai.quantile(0.50),
        nilai.quantile(0.75),
        nilai.max(),
        nilai.var(),
        nilai.std(),
        nilai.sem(),
        nilai.skew(),
        nilai.kurtosis()
    ]
})

display(statistik.round(2))
```
